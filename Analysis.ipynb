```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

df['Visitor Type'] = df['Previous Visits'].apply(
    lambda x: 'New Visitor' if x == 0 else 'Returning Visitor'
)

visitor = df['Visitor Type'].value_counts()

plt.figure(figsize=(6,4))
visitor.plot(kind='bar')

plt.title('New vs Returning Visitors')
plt.xlabel('Visitor Type')
plt.ylabel('Count')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

df['Engagement Score'] = (
    df['Page Views'] * 0.4 +
    df['Session Duration'] * 0.3 +
    df['Time on Page'] * 0.3
)

engagement = df.groupby('Traffic Source')['Engagement Score'].mean()

plt.figure(figsize=(8,5))
engagement.plot(kind='bar')

plt.title('Average Engagement Score by Traffic Source')
plt.xlabel('Traffic Source')
plt.ylabel('Engagement Score')

for i, v in enumerate(engagement):
    plt.text(i, v+0.05, f'{v:.2f}', ha='center')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

session = df.groupby('Traffic Source')['Session Duration'].mean()

plt.figure(figsize=(8,5))
plt.plot(session.index, session.values, marker='o')

plt.title('Average Session Duration by Traffic Source')
plt.xlabel('Traffic Source')
plt.ylabel('Average Session Duration')

for i, v in enumerate(session.values):
    plt.text(i, v, f'{v:.2f}')

plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

# Create a new variable
df['Visitor Type'] = df['Previous Visits'].apply(
    lambda x: 'New Visitor' if x == 0 else 'Returning Visitor'
)

# Count visitors
visitor_counts = df['Visitor Type'].value_counts()

# Line Chart
plt.figure(figsize=(7,5))
plt.plot(visitor_counts.index, visitor_counts.values, marker='o')

plt.title('New vs Returning Visitors')
plt.xlabel('Visitor Type')
plt.ylabel('Number of Visitors')

for i, v in enumerate(visitor_counts.values):
    plt.text(i, v, str(v))

plt.grid(True)
plt.show()
```
